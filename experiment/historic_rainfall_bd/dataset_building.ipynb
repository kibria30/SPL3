```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('customized_daily_rainfall_data.csv', index_col=0)
df.head()
```

```python
df.shape
```

```python
# i want to make a all in one csv cols will be year, month, Day, time, station_1_name(value will be rainfall), station_2_name(value will be rainfall), station_3_name(value will be rainfall) and so on for all the stations.
```

```python
# Convert the long station table into one row per date and one column per station.
rainfall_long = df.reset_index(drop=True).copy()
rainfall_long = rainfall_long.rename(columns={
    "Year": "year",
    "Month": "month",
    "Day": "Day",
    "Station": "station",
    "Rainfall": "rainfall",
})

all_stations_rainfall = (
    rainfall_long.pivot_table(
        index=["year", "month", "Day"],
        columns="station",
        values="rainfall",
        aggfunc="first",
    )
    .reset_index()
    .rename_axis(columns=None)
    .sort_values(["year", "month", "Day"])
    .reset_index(drop=True)
    )

# output_path = "all_stations_daily_rainfall.csv"
# all_stations_rainfall.to_csv(output_path, index=False)

# print(f"Saved {output_path} with shape {all_stations_rainfall.shape}")
display(all_stations_rainfall.head())
```

```python
df = all_stations_rainfall
```

```python
# find columns with with missing value counts
missing_values = df.isnull().sum()
print(missing_values)
```

```python
372/30
```

```python
# drop columns with missing values count > 372
df = df.loc[:, ~(df.isnull().sum() > 372)]
df.head()
```

```python
# find columns with with missing value counts
missing_values = df.isnull().sum()
print(missing_values)
```

```python
columns_with_missing_values = missing_values[missing_values > 0].index.tolist()
print("Columns with missing values:", columns_with_missing_values)
```

```python
# find the missing values year, month, day for station name 'Dhaka'
for station in columns_with_missing_values:
    station_data = df[station]
    station_missing = station_data[station_data.isnull()]
    missing_dates = df.loc[station_missing.index, ['year', 'month', 'Day']]
    print(f"Missing dates for station {station}:")
    print(missing_dates)
```

```python
# also consider these columns for dropping: 'Comilla', 'Jessore', 'Khulna', 'Srimangal']
cols_to_drop = ['Comilla', 'Jessore', 'Khulna', 'Srimangal']
df = df.drop(columns=cols_to_drop)
df.head()
```

```python
# keep rows where the year starts from 1975 also set default index it from 0 to n-1
df = df[df['year'] >= 1975]
df = df.reset_index(drop=True)
df.head()
```

```python
# now count the missing values again
missing_values = df.isnull().sum()
print(missing_values)
```

```python
cleaned_daily_rainfall_data_1975_to_2016 = df
cleaned_daily_rainfall_data_1975_to_2016.to_csv('cleaned_daily_rainfall_data_1975_to_2016.csv', index=False)
```
